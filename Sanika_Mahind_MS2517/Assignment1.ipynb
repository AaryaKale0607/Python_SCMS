## Problem 1: Average Marks and Grade

```python
marks = []
for i in range(5):
    m = float(input(f"Enter marks for subject {i+1}: "))
    marks.append(m)

total = 0
for m in marks:
    total += m
average = total / len(marks)

if average >= 90:
    grade = "A+"
elif average >= 75:
    grade = "A"
elif average >= 60:
    grade = "B"
elif average >= 40:
    grade = "C"
else:
    grade = "Fail"

print("Marks:", marks)
print("Average:", average)
print("Grade:", grade)
```

# Explanation: Marks for five subjects are collected in a loop and stored in a list. The average is computed manually by summing and dividing by count, then compared against threshold ranges to decide the grade.

# Problem 2: Sentence Analysis

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
word_count = len(words)
char_count = len(sentence.replace(" ", ""))

vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0

for ch in sentence:
    if ch.isalpha():
        if ch in vowels:
            vowel_count += 1
        else:
            consonant_count += 1

print("Number of words:", word_count)
print("Number of characters:", char_count)
print("Number of vowels:", vowel_count)
print("Number of consonants:", consonant_count)
```

# Explanation: split() breaks the sentence into words on whitespace. Characters are counted after removing spaces. Each character is checked with isalpha() to skip punctuation, then classified as a vowel or consonant.

# Problem 3: Separate Even and Odd Numbers

```python
n = int(input("How many numbers? "))
numbers = []
for i in range(n):
    numbers.append(int(input(f"Enter number {i+1}: ")))

even_list = []
odd_list = []

for num in numbers:
    if num % 2 == 0:
        even_list.append(num)
    else:
        odd_list.append(num)

print("Even numbers:", even_list)
print("Odd numbers:", odd_list)
```

# Explanation: Each entered number is checked with the modulus operator; if divisible by 2 with no remainder it's appended to the even list, otherwise to the odd list.

# Problem 4: Largest, Smallest, Sum, Average (no built-ins)

```python
n = int(input("How many numbers? "))
numbers = []
for i in range(n):
    numbers.append(int(input(f"Enter number {i+1}: ")))

largest = numbers[0]
smallest = numbers[0]
total = 0

for num in numbers:
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num
    total += num

average = total / len(numbers)

print("Largest:", largest)
print("Smallest:", smallest)
print("Sum:", total)
print("Average:", average)
```

# Explanation: The first element is assumed to be both the largest and smallest initially. As the loop scans each number, it updates largest/smallest whenever it finds a bigger or smaller value, and accumulates the total for the sum and average — without calling max(), min(), or sum().

# Problem 5: Palindrome Check

```python
text = input("Enter a string: ")

cleaned = text.replace(" ", "").lower()
reversed_text = cleaned[::-1]

if cleaned == reversed_text:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

# Explanation: Spaces are removed and the string is lowercased so case and spacing don't affect the comparison. The cleaned string is then compared to its reverse ([::-1]); if they match, it's a palindrome.

# Problem 6: Tuple of Marks

```python
n = int(input("How many students? "))
marks_list = []
for i in range(n):
    marks_list.append(int(input(f"Enter marks for student {i+1}: ")))

marks_tuple = tuple(marks_list)

highest = marks_tuple[0]
lowest = marks_tuple[0]
total = 0

for m in marks_tuple:
    if m > highest:
        highest = m
    if m < lowest:
        lowest = m
    total += m

average = total / len(marks_tuple)

above_average_count = 0
for m in marks_tuple:
    if m > average:
        above_average_count += 1

print("Marks:", marks_tuple)
print("Highest mark:", highest)
print("Lowest mark:", lowest)
print("Students scoring above average:", above_average_count)
```

# Explanation: Marks are stored as a tuple. Highest and lowest are found by scanning, the average is computed manually, and a second pass counts how many marks exceed that average.

# Problem 8: Character Frequency Count

```python
text = input("Enter a string: ")

frequency = {}
for ch in text:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

for ch, count in frequency.items():
    print(f"{ch} : {count}")
```

# Explanation: The dictionary frequency tracks how many times each character appears. For every character, its count is incremented if already seen, or initialized to 1 if it's new.

# Problem 9: Set Operations

```python
set1_input = input("Enter elements of first set (space separated): ")
set2_input = input("Enter elements of second set (space separated): ")

set1 = set(set1_input.split())
set2 = set(set2_input.split())

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference (Set1 - Set2):", set1 - set2)
print("Symmetric Difference:", set1 ^ set2)
```

# Explanation: Two sets are built from user input. The |, &, -, and ^ operators directly compute union, intersection, difference, and symmetric difference respectively.

# Problem 10: Contact Book (Menu-Driven)

```python
contacts = {}

while True:
    print("\n--- Contact Book ---")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Display All Contacts")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter name: ")
        number = input("Enter phone number: ")
        contacts[name] = number
        print("Contact added.")

    elif choice == "2":
        name = input("Enter name to search: ")
        if name in contacts:
            print(f"{name}: {contacts[name]}")
        else:
            print("Contact not found.")

    elif choice == "3":
        name = input("Enter name to update: ")
        if name in contacts:
            number = input("Enter new phone number: ")
            contacts[name] = number
            print("Contact updated.")
        else:
            print("Contact not found.")

    elif choice == "4":
        name = input("Enter name to delete: ")
        if name in contacts:
            del contacts[name]
            print("Contact deleted.")
        else:
            print("Contact not found.")

    elif choice == "5":
        if contacts:
            for name, number in contacts.items():
                print(f"{name}: {number}")
        else:
            print("No contacts saved.")

    elif choice == "6":
        print("Exiting Contact Book.")
        break

    else:
        print("Invalid choice, try again.")
```

# Explanation: A while True loop displays a menu and repeatedly asks for a choice. Each option performs a dictionary operation — adding, searching, updating, deleting, or listing contacts — until the user chooses to exit, which breaks the loop.

# Problem 12: Decreasing Star Pattern

```python
n = int(input("Enter n: "))

for i in range(n, 0, -1):
    print("*" * i)
```

# Explanation: The loop counts down from n to 1, printing one fewer star each row, producing the shrinking triangle.

# Problem 13: Floyd's Triangle

```python
n = int(input("Enter n: "))

num = 1
for i in range(1, n + 1):
    row = ""
    for j in range(i):
        row += str(num) + " "
        num += 1
    print(row.strip())
```

# Explanation: A running counter num keeps increasing across the whole triangle. The outer loop controls the row number, and the inner loop prints that many consecutive numbers in a row before moving to the next.

# Problem 14: Remove Duplicates (Preserve Order)

```python
n = int(input("How many numbers? "))
numbers = []
for i in range(n):
    numbers.append(int(input(f"Enter number {i+1}: ")))

unique_numbers = []
for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)

print("Original list:", numbers)
print("List without duplicates:", unique_numbers)
```

# Explanation: A new list unique_numbers is built by checking, for each number, whether it has already been added; if not, it's appended. Since we scan in original order, the order of first appearances is preserved.

# Problem 15: Student Record Management System

```python
students = {}

while True:
    print("\n--- Student Record Management System ---")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display All Students")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        roll = input("Enter roll number: ")
        name = input("Enter name: ")
        course = input("Enter course: ")
        marks = input("Enter marks: ")
        students[roll] = {"Name": name, "Course": course, "Marks": marks}
        print("Student added.")

    elif choice == "2":
        roll = input("Enter roll number to search: ")
        if roll in students:
            info = students[roll]
            print(f"Roll No: {roll}, Name: {info['Name']}, Course: {info['Course']}, Marks: {info['Marks']}")
        else:
            print("Student not found.")

    elif choice == "3":
        roll = input("Enter roll number to update: ")
        if roll in students:
            name = input("Enter new name: ")
            course = input("Enter new course: ")
            marks = input("Enter new marks: ")
            students[roll] = {"Name": name, "Course": course, "Marks": marks}
            print("Student record updated.")
        else:
            print("Student not found.")

    elif choice == "4":
        roll = input("Enter roll number to delete: ")
        if roll in students:
            del students[roll]
            print("Student record deleted.")
        else:
            print("Student not found.")

    elif choice == "5":
        if students:
            for roll, info in students.items():
                print(f"Roll No: {roll}, Name: {info['Name']}, Course: {info['Course']}, Marks: {info['Marks']}")
        else:
            print("No student records available.")

    elif choice == "6":
        print("Exiting Student Record Management System.")
        break

    else:
        print("Invalid choice, try again.")
```

# Explanation: Each student's roll number is used as a dictionary key, mapping to a nested dictionary holding their name, course, and marks. The same menu-driven pattern as the contact book handles adding, searching, updating, deleting, and displaying records.

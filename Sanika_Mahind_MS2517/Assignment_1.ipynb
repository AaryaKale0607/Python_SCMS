```python
a = 10
b = 20
a, b = b, a
print(a)
print(b)
```

# Explanation: a starts as 10 and b starts as 20. The statement a, b = b, a assigns both new values at the same time, so a becomes 20 and b becomes 10, without needing a temporary variable.

```python
text = "Python Programming"
print(text[0])
print(text[-1])
print(text[0:6])
print(text[::-1])
```

# Explanation: text[0] returns the first character, text[-1] returns the last character using negative indexing, text[0:6] slices out the first six characters, and text[::-1] walks through the string with a step of -1, which reverses it.

```python
numbers = [10, 20, 30]
numbers.append(40)
numbers.remove(20)
print(numbers)
```

# Explanation: append(40) adds 40 to the end of the list, and remove(20) deletes the first occurrence of the value 20, leaving [10, 30, 40].

```python
text = "Programming"
print(text[0:5])
print(text[3:])
print(text[:6])
print(text[::-1])
```

# Explanation: text[0:5] takes characters from index 0 up to (not including) 5. text[3:] takes everything from index 3 to the end. text[:6] takes everything from the start up to index 6. text[::-1] reverses the whole string.

```python
age = input("Enter your age: ")
print(age + age)
age = int(age)
print(age + age)
```

# Explanation: input() always returns a string, so before conversion age + age concatenates the text (e.g. "19") with itself to give "1919". After int(age) converts it to a number, age + age performs numeric addition instead, giving 38.

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(fruits[0])
print(fruits[-1])
print(fruits[1:3])
```

# Explanation: fruits[0] accesses the first item, fruits[-1] accesses the last item using negative indexing, and fruits[1:3] slices out the items at index 1 and 2.

```python
numbers = [10, 20, 30]
numbers[1] = 200
print(numbers)
```

# Explanation: The list starts as [10, 20, 30]. Assigning numbers[1] = 200 replaces the element at index 1, giving [10, 200, 30].

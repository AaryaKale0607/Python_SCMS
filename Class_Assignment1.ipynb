## Problem - 1

```python
marks = []

for i in range(5) :
    mark = int(input("Enter marks for subject {} : ".format(i + 1)))
    marks.append(mark)

total = 0 

for mark in marks :
    total += mark
average = total / 5

if average >= 90:
    grade = "A+"
elif average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 50:
    grade = "D"
else:
    grade = "F"

print("Marks : " , marks)
print("Average : " , average)
print("Grade : " , grade)
```

## Problem - 2

```python
sentence = input("Enter a sentence : ")

words = sentence.split()

word_count = len(words)
char_count = 0 
vowel_count = 0
consonant_count = 0

for ch in sentence:
    if ch != " ":
        char_count += 1

        if ch.lower() in "aeiou" :
            vowel_count += 1
        elif ch.isalpha():
            consonant_count += 1

print("Number of words : " , word_count)
print("Number of characters : " , char_count)
print("Number of vowels : " , vowel_count)
print("Number of consonants : " , consonant_count) 
```

## Problem - 3

```python
numbers = input("Enter integers separated by spaces :").split()

even = []
odd = []

for value in numbers:
    number = int(value)

    if number % 2 == 0:
        even.append(number)
    else:
        odd.append(number)

print("Even numbers : " , even)
print("Odd numbers : " , odd)

```

## Problem - 4

```python
numbers = input("Enter integers separated by spaces : ").split()

numbers = [int (x) for x in numbers ]

#Initialize
largest = numbers[0]
smallest = numbers[0]

total = 0

#Find largest , smallest and sum

for number in numbers:
    if number > largest:
        largest = number

    if number < smallest:
        smallest = number

    total += number

average = total / len(numbers)

print("Largest number : ", largest)
print("Smallest number :", smallest)
print("Sum :", total)
print("Average :" , average)
```

## Problem - 5

```python
string = input("Enter a string : ")

#Remove spaces and convert to lower case
cleaned = ""

for ch in string:
    if ch != " ":
        cleaned += ch.lower()

#Check palindrome
reverse = ""

for i in range (len(cleaned) - 1, -1 , -1):
    reverse += cleaned [i]

if cleaned == reverse:
    print("The string is a Palindrome.")
else:
    print("The string is not Palindrome.")
```

## Problem - 6

```python
marks_input = input("Enter sudent marks separated by spaces : ")

marks_list = []

for value in marks_input.split():
    marks_list.append(float(value))

marks = tuple(marks_list)

#Find highest and lowest

highest = marks[0]
lowest = marks[0]

total = 0

for mark in marks:
    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

    total += mark

average = total / len(marks)

# count students above average

above_average = 0

for mark in marks:
    if mark > average:
        above_average += 1

print("Marks : " , marks)
print("Highest : " , highest)
print("Lowest : " , lowest)
print("Average : " , average)
print("Students above average : " , above_average)
```

## Problem - 9

```python
set1_input = input("Enter elements of first set : ").split()
set2_input = input("Enter elements of second set : ").split()

set1 = set(set1_input)
set2 = set(set2_input)

print("Set 1 : ", set1)
print("Set 2 : ", set2)

print("Union : " , set1 | set2)
print("Intersection : " , set1 & set2)
print("Difference(Set 1 - Set 2) : " , set1 - set2)
print("Symmetric Difference : " , set1 ^ set2 )
```

## Problem - 10

```python
contacts = {}

while True:
    print("\n--------CONTACT BOOK --------")
    print("1.Add Contact")
    print("2.Search Contact")
    print("3.Update Contact")
    print("4.Delete Contact")
    print("5.Display All Contacts")
    print("6.Exit")

    choice = input("Enter your choice (1 - 6) : ") 

    #Add Contact
    if choice == "1":
        name = input("Enter name : ")
        phone = input("Enter phone number : ")

        contacts[name] = phone

        print("Contact added successfully.")

    #Search Contact
    elif choice == "2":
        name = input("Enetr name to search : ")

        if name in contacts:
            print("Phone : " , contacts[name])
        else:
            print("Contatc not found")

    #Update Contact
    elif choice == "3":
        name = input("Enter name to update : ")

        if name in contacts:
            phone = input("Enter new phone number : ")  
            contacts[name] = phone

            print("Contact updated successfully.")
        else:
            print("Contact not found.")

    #Delete Contact
    elif choice == "4":
        name = input("Enter name to delete : ")

        if name in contacts:
            del contacts[name]
            print("Contact deleted successfully .")
        else:
            print("Contact not found.")

    #Display contacts
    elif choice == "5":
        if len(contacts) == 0:
            print(" No contacts available.")
        else:
            print("\nContacts : ")

            for name in contacts:
                print(name , ":" , contacts[name])

    #Exit
    elif choice == "6":
        print("Exiting Contact Book.")
        break

    else:
        print("Invalid choice.")      
```

## Problem - 15

```python
students = {}

while True:

    print("\n===== STUDENT RECORD MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display All Students")
    print("6. Exit")

    choice = input("Enter your choice: ")

    # Add student
    if choice == "1":

        roll = input("Enter roll number: ")

        if roll in students:
            print("Student with this roll number already exists.")
        else:
            name = input("Enter name: ")
            course = input("Enter course: ")
            marks = float(input("Enter marks: "))

            students[roll] = {
                "name": name,
                "course": course,
                "marks": marks
            }

            print("Student added successfully.")

    # Search student
    elif choice == "2":

        roll = input("Enter roll number to search: ")

        if roll in students:
            print("\nStudent Details")
            print("Roll Number:", roll)
            print("Name:", students[roll]["name"])
            print("Course:", students[roll]["course"])
            print("Marks:", students[roll]["marks"])
        else:
            print("Student not found.")

    # Update student
    elif choice == "3":

        roll = input("Enter roll number to update: ")

        if roll in students:

            print("\nEnter new information")

            name = input("Enter new name: ")
            course = input("Enter new course: ")
            marks = float(input("Enter new marks: "))

            students[roll]["name"] = name
            students[roll]["course"] = course
            students[roll]["marks"] = marks

            print("Student information updated successfully.")

        else:
            print("Student not found.")

    # Delete student
    elif choice == "4":

        roll = input("Enter roll number to delete: ")

        if roll in students:
            del students[roll]
            print("Student deleted successfully.")
        else:
            print("Student not found.")

    # Display all students
    elif choice == "5":

        if len(students) == 0:
            print("No student records available.")

        else:
            print("\n===== ALL STUDENTS =====")

            for roll in students:

                print("\nRoll Number:", roll)
                print("Name:", students[roll]["name"])
                print("Course:", students[roll]["course"])
                print("Marks:", students[roll]["marks"])

    # Exit
    elif choice == "6":

        print("Exiting Student Record Management System.")
        break

    else:

        print("Invalid choice. Please try again.")
```

## Problem 2 - Temperature Analysis

```python
import numpy as np

def temperature_analysis(temperatures):
    avg = np.mean(temperatures)
    maximum = np.max(temperatures)
    minimum = np.min(temperatures)
    highest_day = np.argmax(temperatures) + 1
    above_avg = temperatures[temperatures > avg]

    return avg , maximum , minimum , highest_day , above_avg

temperatures = np.array([32.5, 31.8, 30.2, 33.1, 34.5, 29.8, 31.0])

avg , maximum, minimum, highest_day , above_avg = \
temperature_analysis(temperatures)

print(f"Temperatures: {temperatures}")
print(f"Average temprature: {avg}")
print(f"Maximum temperature: {maximum}")
print(f"Minimum temperature: {minimum}")
print(f"Day with highest temperature: {highest_day}")
print(f"Tempratures above average: {above_avg}")
```

## Problem 5 - Numbers Above a Threshold

```python
import numpy as np

np.random.seed(10)

numbers = np.random.randint(1 , 101 , 10)

greater_than_50 = numbers[numbers > 50]
between_25_and_75 = numbers[(numbers >= 25) & (numbers <= 75)]

count = np.sum(numbers > 50)

average = np.mean(greater_than_50)

print(f"Generated Array:\n {numbers}")

print(f"\nValues greater than 50:\n {greater_than_50}")

print(f"\nValues between 25 and 75:\n {between_25_and_75}")

print(f"\nCount of values greater than 50: {count}")

print(f"\nAverage of values greater than 50: {average}")
```

## Problem 8 - Scientific Data Transformation

```python
import numpy as np

def transform_data(data):
    square = data **2
    square_root = np.sqrt(data)
    exponential = np.exp(data)

    return square, square_root , exponential

data = np.array([2, 4, 6, 8, 10, 12])

square, square_root, exponential = transform_data(data)

print(f"Original data:\n {data}")

print(f"\nSquare:\n {square}")

print(f"\nSquare root:\n {square_root}")

print(f"\nExponential:\n {exponential}")

print(f"\nMean of square: {np.mean(square)}")

print(f"Mean of square root: {np.mean(square_root)}")

print(f"Mean of exponential: {np.mean(exponential)}")
```

## Problem 13 - Experiment Comaprison

```python
import numpy as np 

exp_A = np.random.randint(10 , 20 , 7)

exp_B = np.random.randint(10 , 20 , 7)

print(f"Experiment A : {exp_A}")
print(f"\nExperiment B : {exp_B}")

print(f"\nMean of Experiment A : {np.mean(exp_A)}")
print(f"\nMean of Experiment B : {np.mean(exp_B)}")

print(f"\nStandard Deviation of Experment A : {np.std(exp_A)}")
print(f"\nStandard Deviation of experiment B : {np.std(exp_B)}")

print(f"\nMaximum of Experment A : {np.max(exp_A)}")
print(f"\nMaximum of Experment B : {np.max(exp_B)}")

print(f"\nMinimum of Experment A : {np.max(exp_A)}")
print(f"\nMinimum of Experment B : {np.max(exp_B)}")

print(f"\nDifference between  Experment A & Experiment B : {exp_A - exp_B}")

if np.std(exp_A) > np.std(exp_B):
    print("\nExperiment A  shows greater variation.")
else:
    print("\nExperiment B  shows greater variation.")
```

## Problem 15 - Final Student Performance Analysis

```python
import numpy as np

def generate_marks() :
    np.random.seed(7)
    return np.random.randint(0 ,101 , (10 , 4))

def calculate_grade(average):
    if average >= 90 :
        return "A+"
    elif average >= 80 :
        return "A"
    elif average >= 70 :
        return "B"
    elif average >= 60 :
        return "C"
    elif average >= 50 :
        return "D"
    else :
        return "F"

def analyze_stu(marks):
    stu_average = np.mean(marks, axis=1)
    sub_average = np.mean(marks , axis=0)

    highest_stu = np.argmax(stu_average) + 1
    lowest_stu = np.argmin(stu_average) + 1

    higest_marks = np.max(marks)
    lowest_marks = np.min(marks)

    stu_above_75 = np.sum(stu_average > 75)

    marks_above_90 = marks[marks > 90]

    grades = []

    for average in stu_average:
        grades.append(calculate_grade(average))

    stu_info = []

    for i in range(10):
        stu_info.append({
            "Student" : i + 1,
            "Average" : stu_average[i],
            "Grade" : grades[i]
        })

    print(f"\nStudents marks :\n {marks}")

    print(f"\nAverage mark of every student :\n {stu_average}")

    print(f"\nAverage mark of every subject :\n {sub_average}")

    print(f"\nStudent with highest average : {highest_stu}")

    print(f"\nStudent with lowest average : {lowest_stu}")

    print(f"\nHighest marks : {higest_marks}")

    print(f"\nLowest marks : {lowest_marks}")

    print(f"\nStudents with average above 75 :\n {stu_above_75}")

    print(f"\nMarks greater than 90 :\n {marks_above_90}")

    print("\nFinal Student Information : ")

    for stu in stu_info:
        print(stu)

marks = generate_marks()
analyze_stu(marks)
```

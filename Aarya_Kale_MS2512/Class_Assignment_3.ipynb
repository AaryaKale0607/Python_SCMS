## Problem - 7 --- Create a Bank Account Class

```python
class BankAccount:

    def __init__(self, acc_num, owner, balance): 
        self.acc_num = acc_num 
        self.owner = owner 
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            return True
        else:
            return False

acc1 = BankAccount(1 , " Aarya " , 1000)
acc2 = BankAccount(2 , " Kartik" , 2000)

print("----Bank Accounts----")
print(f"\nAccount Number : {acc1.acc_num} , \nOwner : {acc1.owner} , \nBalance : {acc1.balance}")
print(f"\nAccount Number : {acc2.acc_num} , \nOwner : {acc2.owner} , \nBalance : {acc2.balance}")
```

## Problem - 8 --- Bank Account Methods

```python
acc = BankAccount(3 , " Bob " , 8000)

print(f"\nInitial Balance for test account : {acc.balance}")

acc.deposit(2000)
print(f"\nBalance after deposit : {acc.balance}")

acc.withdraw(500)
print(f"\nBalance after withdrawal : {acc.balance}")
```

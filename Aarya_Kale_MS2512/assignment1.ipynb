## Experiment 7 - List Mutability

```python
numbers = [10 , 20 , 30]

numbers[1] = 200

print(numbers)
```

Initially  list contains [10, 20, 30]

numbers[1] = 200 changes  element at index 1 from 20 to 200

the final list becomes [10, 200, 30]

A list can be modified because lists in Python are **mutable**, which means their elements can be changed after the list is created.

Strings, on the other hand, are **immutable**, which means individual characters of a string cannot be changed after the string is created.

Hence, list elements can be updated directly, while string characters cannot be modified directly.
